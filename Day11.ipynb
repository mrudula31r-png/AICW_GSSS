```python
 ## Helper function to display images in Jupyter Notebook
def display_image(img, title="Image", cmap_type='gray'):
    plt.figure(figsize=(10, 8))
    plt.imshow(img, cmap=cmap_type)
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
# Read an image using OpenCV
img = cv2.imread('Couple.jpg')  # replace 'your_image_path.jpg' with your image file path

# Convert the image from BGR to RGB format
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the image using Matplotlib
display_image(img_rgb, title="Original Image", cmap_type=None)
```

```python
# 2. Resize Image and Display
# Resize the image
img_resized = cv2.resize(img_rgb, (200, 200))

# Display the resized image
display_image(img_resized, title="Resized Image", cmap_type=None)
```

```python
# 3. Image Annotations
# Draw a rectangle on the image
img_annotated = img_rgb.copy()
cv2.rectangle(img_annotated, (50, 50), (150, 150), (255, 0, 0), 3)  # BGR color format, here it is blue

# Draw a circle on the image
cv2.circle(img_annotated, (100, 100), 30, (0, 255, 0), 3)  # BGR color format, here it is green

# Add text to the image
cv2.putText(img_annotated, 'Hello', (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)  # BGR color format, here it is red

# Display the annotated image
display_image(img_annotated, title="Annotated Image", cmap_type=None)
```

```python
# 4. Image Thresholding
# Convert the image to grayscale
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Apply binary thresholding
_, img_thresh = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

# Display the thresholded image
display_image(img_thresh, title="Thresholded Image")
```

```python

# 5. Edge Detection
# Apply Canny edge detection
edges = cv2.Canny(img_gray, 100, 200)

# Display the edges
display_image(edges, title="Edge Detection")
```

```python
# sample sentence
text_sample='''He determined to drop his litigation with the monastry, and relinquish his claims to the wood cutting and fishery rights at once . He was more ready to do this because the rights ahad become much lesss valuable, and he had indeed the vaguest idea where the wood and river in questions were. '''
# set the stop words
stop_words = set(stopwords.words('english'))
# tokens
word_tokens = word_tokenize(text_sample)
filtered_sentence =[]
for w in word_tokens:
    if w not in stop_words:
        filtered_sentence.append(w)

print('Original Sentence')
print(" ".join(word_tokens))
print('*'*50)
print('Filtered Sentence')
print(" ".join(filtered_sentence))
```

```python
from collections import Counter
from string import punctuation
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')


def get_hotwords(text):
    result = []
    pos_tag = ['PROPN', 'ADJ', 'NOUN']
    doc = nlp(text.lower())

    for token in doc:
        if (token.text in nlp.Defaults.stop_words or token.text in punctuation):
            continue

        if (token.pos_ in pos_tag):
            result.append(token.text)

    return result
```

```python
new_text = """When it comes to evaluating the performance of keyword extractors, you can use some of the standard metrics in Machine Learning: accuracy, precision, recall and F1 Score. However, these metrics don't reflect partial matches. They only consider the perfect match between the extracted segments and the correct prediction of the tag"""
```

```python
output = set(get_hotwords(new_text))
most_common_list = Counter(output).most_common(10)

for item in most_common_list:
    print(item[0])
```

```python
# Lemmatization - normalization - dict
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
# Explanation of 'df' in NLP context:
# 'df' in the previous cells refers to a pandas DataFrame containing the Document-Term Matrix.
# It represents the frequency of terms (words) in a collection of documents.
# Each row corresponds to a document, and each column corresponds to a unique term.
# The values in the DataFrame are the counts of each term in each document.
```

```python
Porter = PorterStemmer()
lemmatizer = WordNetLemmatizer()
print(f'better\n Stemming: {Porter.stem("better")}\n Lemmatization: {lemmatizer.lemmatize("better", pos="a")}')
```
