## Implementing RNN Classifier for IMDB Dataset

```python
# importing nec libs
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN, Embedding
from tensorflow.keras.preprocessing import sequence
tf.random.set_seed(7) # setting seed value for reproducibility
```

```python
# load the dataset & keep the top n words
top_words = 5000
(xtr, ytr), (xte, yte) = imdb.load_data(num_words=top_words)
```

```python
# preprocess the data
max_review_length = 500
xtr = sequence.pad_sequences(xtr, maxlen=max_review_length)
xte = sequence.pad_sequences(xte, maxlen=max_review_length)
```

```python
# create RNN model & train
embedding_vector_length = 32
model = Sequential()
model.add(Embedding(top_words, embedding_vector_length, input_length=max_review_length))
model.add(SimpleRNN(100))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
print(model.summary())
model.fit(xtr, ytr, validation_data=(xte, yte), epochs=5, batch_size=64)
```

```python
# importing nec libs
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN, Embedding, Input
from tensorflow.keras.preprocessing import sequence
tf.random.set_seed(7) # setting seed value for reproducibility
```

```python
# load the dataset & keep the top n words
top_words = 5000
(xtr, ytr), (xte, yte) = imdb.load_data(num_words=top_words)
```

```python
# preprocess the data
max_review_length = 500
xtr = sequence.pad_sequences(xtr, maxlen=max_review_length)
xte = sequence.pad_sequences(xte, maxlen=max_review_length)
```

```python
# create RNN model & train
embedding_vector_length = 32
model = Sequential([
    Input(shape=(max_review_length,)),
    Embedding(input_dim=top_words, output_dim=embedding_vector_length),
    SimpleRNN(100),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam',
             metrics=['accuracy'])
```

```python
print(model.summary())
```

```python
model.fit(xtr, ytr, validation_data=(xte, yte), epochs=3, batch_size=64)
```

```python
score = model.evaluate(xte, yte, verbose=0)
print(score[0] * 100, score[1]*100)
```
