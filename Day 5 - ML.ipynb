```python
from sklearn.model_selection import train_test_split , cross_val_score , GridSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
import sklearn.metrics as mt
import xgboost as xgb
import category_encoders as ce
from sklearn.preprocessing import MinMaxScaler
```

```python
hr.columns
```

```python
hr=hr.drop(['Floor','Area Locality'],axis=1)
hr.columns
```

```python
# Data splitting - 100% - 80 % training set and 20% testing set
X = hr.drop(columns='Rent')                       # Independent variable means taking all
y = hr['Rent']                                    # dependent variable - labeled - which to remove



```

```python
X.head()
```

```python
# Encoding - LOO encoder
encoder = ce.LeaveOneOutEncoder(return_df=True)
```

```python
X = encoder.fit_transform(X , y)
```

```python
X.head()
```

```python
# Standardization - scaling
scalar = MinMaxScaler()           # obj=class
X = scalar.fit_transform(X)         # chaning independent values
# y= scaler.transform()
```

```python
X
```

```python
X_df = pd.DataFrame(X)
X_df.head()
```

```python
# train test split func
X_train , X_test , y_train ,y_test = train_test_split(X , y,
                                                      train_size=0.8,               # 80% for train and 20 for test
                                                      random_state=42)               #randomly set is divided where 42- seed value , it can be anything
```

```python
X_train.shape, X_test.shape
```

```python
# train
lr = LinearRegression()              #obj=class
```

```python
model = lr.fit(X_train , y_train)
```

```python
model
```

```python
# pred
y_pred = model.predict(X_test)
y_pred
```

```python
# evaluate the model
mse=mt.mean_squared_error(y_test, y_pred)                                   # mse =mean squared error
mae = mt.mean_absolute_error(y_test, y_pred)                                 # mean absolute error           
r2 =mt.r2_score(y_test, y_pred)                                             # R Squared score
print(f'The mean squared error is {mse}')                      # mse,mae tells variance  we also have rmse=root mse
print(f'The mean absolute error is {mae}')                      # r2 gives the percentage
print(f'The R squared Score is {r2}')                       # error is also called as residuals
```

```python
# decision tree  -- crrates one treee wid one node
dtr = DecisionTreeRegressor()
model_dtr = dtr.fit(X_train,y_train)
dtr_pred=model_dtr.predict(X_test)
dtr_pred
```

```python
# evaluate the model
mse_dtr=mt.mean_squared_error(y_test, dtr_pred)                                   # mse =mean squared error
mae_dtr = mt.mean_absolute_error(y_test, dtr_pred)                                 # mean absolute error           
r2_dtr =mt.r2_score(y_test, dtr_pred) 
print(f'The mean squared error is {mse_dtr}')                      # mse,mae tells variance  we also have rmse=root mse
print(f'The mean absolute error is {mae_dtr}')                      # r2 gives the percentage
print(f'The R squared Score is {r2_dtr}')
```

```python
rfr= RandomForestRegressor(n_estimators=100)
model_rfr = rfr.fit(X_train, y_train)
rfr_pred = model_rfr.predict(X_test)
rfr_pred
```

```python
# dataset splitting
x=wdf.drop(columns=['quality'])                 # x is independent
Y=wdf['quality']                                 # y is dependent
```

```python
from imblearn.over_sampling import SMOTE
oversample = SMOTE(k_neighbors=4)
x,Y= oversample.fit_resample(x.fillna(0),Y)
Y.value_counts()
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
```

```python
def classify(model,x,Y):
    x_train,x_test,Y_train,Y_test = train_test_split(x,Y,test_size=0.2,
                                                    random_state=9)
    model.fit(x_train,Y_train)
    return model.score(x_test,Y_test)* 100
```

```python
#Applying logistec regression
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
LR_acc = classify(model, x, Y)
LR_acc
```

```python
# Applying RFC
from sklearn.ensemble import RandomForestClassifier
model_rf = RandomForestClassifier()
rf_acc = classify(model_rf,x,Y)
rf_acc
```
